## **2. JSONL creation**

### **Result format**

For the correct QLoRA training, the followin JSONL must be created.

> Done with the **train**, **val** and **test** datasets

---

```jsonl
{
    {
        "diagnosticos": [
            {
                "diagnostico_extraido": "Fractura vertebral",
                "codigo_CIE10": "S12.8"
            },
            {
                "diagnostico_extraido": "Fractura costal",
                "codigo_CIE10": "S22.3"
            },
        ]
    },
    {
        "diagnosticos": [
            {
                "diagnostico_extraido": "Diabetes Mellitus tipo II",
                "codigo_CIE10": "E11.9"
            },
        ]
    }
}
```
---

This JSONL is the combination of all the outputs that must be obtained with each input.

> Note that we *(MrDoc)* want to extract also `pertenencia`, `fecha_alta` and `fecha_baja`, but CodiEsp do not have that values for training purposes

With this, we can train and know how similar the results are to the output wanted

## **Train and val format**

For the LLM QLoRA to be able to learn how to do the Information Extraction (IE) as wanted, the training must be done with the following input and output

- **Input**
    - `instruction` is the resume of what it must do. A **quick system prompt**
    - `guideline` is the **explained system prompt**
    - `schema` is the json format it must follows. Is the **json schema**
    - `input` is the **user prompt**

---

```txt
### INSTRUCTION
{instruction}

### GUIDELINE
{guideline}

### SCHEMA
{schema}

### INPUT
{input}

### OUTPUT
```
---

- **Output**
    - `output` is just a line of the JSONL created previously
---

```txt
{output}
```
---

> The train is done introducing `input + output` as a single prompt, in order for the training to learn. The val process only introduces the `input` and compares the `output` generated with the real one

````python
"""
The JSONL created will have an extra field in each line with the ID of each text for a better relation for the latter input prompt creation.
This field is not going to be introduced in the JSONL for training and evaluation.
"""
data_path = "CodiEsp/Original/final_dataset_v4_to_publish"
column_names = ["archivo_origen", "tipo", "codigo_CIE10", "diagnostico_extraido", "localizacion"]

train_df = pd.read_csv(data_path+"/train/trainX.tsv", sep="\t", header=None, names=column_names)
dev_df = pd.read_csv(data_path+"/dev/devX.tsv", sep="\t", header=None, names=column_names)
test_df = pd.read_csv(data_path+"/test/testX.tsv", sep="\t", header=None, names=column_names)
````

````python
def obtener_jsonl(df):
    jsonl = []
    for row in df[df["tipo"] != "PROCEDIMIENTO"].groupby("archivo_origen"):
        jsonl.append({"diagnosticos": row[1][["archivo_origen", "codigo_CIE10", "diagnostico_extraido"]].to_dict(orient="records")})
    return jsonl


train_jsonl = obtener_jsonl(train_df)
dev_jsonl = obtener_jsonl(dev_df)
test_jsonl = obtener_jsonl(test_df)
````

````python
def save_jsonl(jsonl, path):
    path.parent.mkdir(parents=True, exist_ok=True)
    with open(path, "w", encoding="utf-8") as f:
        for row in jsonl:
            f.write(json.dumps(row, ensure_ascii=False) + "\n")

jsonl_path = Path("CodiEsp/JSONL")

save_jsonl(train_jsonl, jsonl_path / "train.jsonl")
save_jsonl(dev_jsonl, jsonl_path / "dev.jsonl")
save_jsonl(test_jsonl, jsonl_path / "test.jsonl")
````
