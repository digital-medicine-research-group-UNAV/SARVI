```python
import pandas as pd
from itertools import product
from tqdm.auto import tqdm

from SARVI.services.common.ner_funcs import classification_report_ner, create_ultimate_df_ann
from SARVI.data_io.reader import read_ann_list, read_json_single

from SARVI.config import (
    paths
)
```

## **True**

```python
dict_ann_test_true = read_ann_list(paths.data_input / "OpenEvidence")
df_ann_test_true = pd.DataFrame([{"archivo_origen": file_name, **ann_data}for file_name, ann_data in dict_ann_test_true.items()])

label2id = read_json_single(paths.docs_dir / "NERTraining/Span/label2id_span_nerclassifier.json")
```

## **BIO Results**

```python
dict_ann_test_bio = read_ann_list(paths.data_intermediate / "OpenEvidence/ANN/NER" / "BIO")
df_ann_test_bio = pd.DataFrame([{"archivo_origen": file_name, **ann_data}for file_name, ann_data in dict_ann_test_bio.items()])
```

```python
classification_report_ner(df_ann_test_true, df_ann_test_bio, [k for k in label2id.keys() if k != "O"], print_report=True)
```

## **Span Results**

```python
dict_ann_test_span = read_ann_list(paths.data_intermediate / "OpenEvidence/ANN/NER" / "Span")
df_ann_test_span = pd.DataFrame([{"archivo_origen": file_name, **ann_data}for file_name, ann_data in dict_ann_test_span.items()])
```

```python
classification_report_ner(df_ann_test_true, df_ann_test_span, [k for k in label2id.keys() if k != "O"], print_report=True)
```

## **Mix Results**

```python
comb = {}

for overlapped_entities, extend_entities, add_extra_entities in tqdm(product([False, True], repeat=3), total=2**3):
    ultimate_df_ann = create_ultimate_df_ann(principal_df_ann=df_ann_test_bio, secondary_df_ann=df_ann_test_span, overlapped_entities=overlapped_entities, extend_entities=extend_entities, add_extra_entities=add_extra_entities)
    print(f"\n\noverlapped_entities: {overlapped_entities} \\ extend_entities: {extend_entities} \\ add_extra_entities: {add_extra_entities}\n")
    results = classification_report_ner(df_ann_test_true, ultimate_df_ann, [k for k in label2id.keys() if k != "O"], print_report=True)
    comb[overlapped_entities, extend_entities, add_extra_entities] = results["report"]["macro avg"]["f1-score"]

comb
```
