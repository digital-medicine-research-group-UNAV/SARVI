```python
import torch

from MrDoc.models.schemas import PipelineContext, LLMConfig
from MrDoc.config import paths
from MrDoc.main import run
```

```python
llm_config = LLMConfig(
    service="vllm",
    model="openai/gpt-oss-20b",
    device="cuda" if torch.cuda.is_available() else "cpu"
)

ctx = PipelineContext(
    ussage="deterministic",
    folder_and_archive_name="DETERMINISTAS",
    llm_config=llm_config,
    paths=paths,
    cie_10_version="2026",
    json_parse=False,
    device = "cuda" if torch.cuda.is_available() else "cpu"
)
```

```python
tarea = "docx_to_jsons"
# tarea = "jsons_to_xlsx"

modo = "sync"
# modo = "async"

results = run(ctx, tarea, ctx.ussage, modo, ctx.folder_and_archive_name)
```
