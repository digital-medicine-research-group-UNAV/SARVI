## **Sleep mode**

Allow the model to enter a sleep mode where some **resources are liberated** and fast resume when needed

- **Level 1**
    - KEEP model weights (CPU) and ELIMINATE KV cache (vRAM)
- **Level 2**
    - KEEP model buffers (CPU) and ELIMINATE model weights (CPU) and ELIMINATE KV cache (vRAM)

> For more info click [here](https://docs.vllm.ai/en/stable/features/sleep_mode/)

---

## **Startup optimization**

- **Level 0**
    - None
- **Level 1**
    - *Startup*: Moderate startup time
    - *Performance*: Inductor compilation, CUDAGraphMode.PIECEWISE
    - *Use case*: Balance for most development scenar
- **Level 2** - *Default*
    - *Startup:* Longer startup time
    - *Performance*: -O1 + CUDAGraphMode.FULL_AND_PIECEWISE
    - *Use case*: Production workloads where performance is important. This is the default use case. It is also very similar to the previous default. The primary difference is that noop & fusion flags are enabled.

> For more info click [here](https://docs.vllm.ai/en/stable/design/optimization_levels/)

````python
from vllm import LLM, SamplingParams
from vllm.sampling_params import StructuredOutputsParams
from vllm.distributed import destroy_distributed_environment, destroy_model_parallel, cleanup_dist_env_and_memory

import gc
import json
import torch
import contextlib
import torch.distributed as dist

import os
from pathlib import Path
````

### **Reproducibility**

Two types of env variables

- `VLLM_BATCH_INVARIANT` = 1

    - **BEST but still experimental**. Not possible because the compute capability must be >= 9.0 and our machine *(digimed)* is 8.9
    > *For calculation* 
    > ```bash
    > nvidia-smi --query-gpu=compute_cap --format=csv,noheader
    > ```

- `VLLM_ENABLE_V1_MULTIPROCESSING` = 0

    - **ONLY if vLLM used in Offline mode, i.e. not API ussage**. Makes the KV blocks creation deterministic.

Even with the above settings, vLLM only provides reproducibility when it runs on the same hardware and the same vLLM version.

This is used aside of using the classical `temperature=0`, which must also be used

> For more info click [here](https://docs.vllm.ai/en/stable/usage/reproducibility/)

````python
# os.environ["VLLM_BATCH_INVARIANT"] = "1"
os.environ["VLLM_ENABLE_V1_MULTIPROCESSING"] = "0"
````

````python
def del_llm(llm):
    del llm
    cleanup_dist_env_and_memory()
    destroy_model_parallel()
    destroy_distributed_environment()

    with contextlib.suppress(Exception):
        dist.destroy_process_group()

    gc.collect()
    torch.cuda.empty_cache()
````

### **Supported models**

1. Supported models **natively by vLLM**

    - Best models in order to make full use of vLLM

    - [**Text models**](https://docs.vllm.ai/en/stable/models/supported_models/#list-of-text-only-language-models) and [**Multi-modal models**](https://docs.vllm.ai/en/stable/models/supported_models/#list-of-multimodal-language-models)

2. **Transformers models**

    - Drop performance of a native vLLM model in a 5%

    - Works for the following:

        - **Modalities**: embedding models, language models and vision-language models*
        - **Architectures**: encoder-only, decoder-only, mixture-of-experts
        - **Attention types**: full attention and/or sliding attention

3. **Custom models**

    - Must be written with **Transformers compatibility** or **native vLLM compatibility**

    - It also can be located inside **HiggingFace Model Hub**

    - To learn **how to construct a model in vLLM click [here](https://docs.vllm.ai/en/stable/contributing/model/)**

> For more info click [here](https://docs.vllm.ai/en/stable/models/supported_models/)

````python
# Sample prompts.
prompts = [
    "Hello, my name is",
    "The president of the United States is",
    "The capital of France is",
    "The future of AI is",
]

# Create a sampling params object.
# json_path = Path(__file__).parent.parent / "docs" / "esquema_selected_and_decider.json"
# with json_path.open(encoding="utf-8") as f:
#     json_schema = json.load(f)

sampling_params = SamplingParams(temperature=0,
                                # extra_args={"your_custom_arg_name": 67}                       # To pass extra args in case any LLM use them
                                # structured_outputs=StructuredOutputsParams(json=json_schema)    # To be able to output directly the json required
                                )

# Create an LLM.
llm = LLM(model="openai/gpt-oss-20b", gpu_memory_utilization=0.4,
        enable_prefix_caching=True,                 # Allow for reusing KV pairs (for repetition of same system prompt) -> Imrpove speed
        # cpu_offload_gb=10,                        # GB to load in CPU (Added to the total of GPU ussage)
        # tensor_parallel_size=1,                   # Used for parallelizing in different GPUs - As digimed only have 1, parallel us 1 (or not setted and default)
        limit_mm_per_prompt={                       # If ussing a multi-modal LLM, we can set the maximum input type parameters and save memory as there is no pre-allocation
            "image": {"count": 0}, 
            "video": {"count": 0}
            },                                      
        # max_num_batched_tokens=16384,             # Allow for a faster Time To First Token (TTFT) as improves decoding parallelism in same GPU. It is activated by default, no need to personal-tunning
                                                    # For more info (https://docs.vllm.ai/en/stable/configuration/optimization/#performance-tuning-with-chunked-prefill)
        # optimization_level=2,                     # Optimization level. As 2 is default (and biggest), it is not necesarry to imply manually
        runner="generate",                          # Declare previuously that it is only to be used in generation process (and not pooling)
        enable_sleep_mode=True                      # To use it just ```llm.slep(leve=1)  /  llm.wake_up()```
        ) 
````

````python
# Generate texts from the prompts.
# The output is a list of RequestOutput objects
# that contain the prompt, generated text, and other information.
outputs = llm.generate(prompts, sampling_params)
# Print the outputs.
print("\nGenerated Outputs:\n" + "-" * 60)
for output in outputs:
    prompt = output.prompt
    generated_text = output.outputs[0].text
    print(f"Prompt:    {prompt!r}")
    print(f"Output:    {generated_text!r}")
    print("-" * 60)
````
